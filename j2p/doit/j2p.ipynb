***
#### 계산기 프로그램을 만들며 클래스 알아보기

```python
class Calculator:
	def __init__(self):
		self.result = 0

	def add(self, num):
		self.result += num
		return self.result
	
	def sub(self, num):
		self.result -= num
		return self.result
```

```python
Cal1 = Calculator()

print(Cal1.add(3))
print(Cal1.add(4))
print(Cal1.sub(3))
print(Cal1.sub(7))
```

### 클래스와 객체
***
### 사직연산 클래스 만들기
***
#### 클래스를 어떻게 만들지 구상하기
#### 객체에 연산할 숫자 지정하기

```python
a = FourCal()
a.setdata(4, 2)
print(a.fst)
print(a.snd)

b = FourCal()
b.setdata(3, 7)
print(b.fst)
print(b.snd)
print("a.fst ->", a.fst)

print('a.add() ->', a.add())
```

```python
class FourCal:
	def setdata(self, fst, snd):
		self.fst = fst
		self.snd = snd

	def add(self):
		result = self.fst + self.snd
		return result
	
	def sub(self):
		result = self.fst - self.snd
		return result
	
	def mul(self):
		result = self.fst * self.snd
		return result
	
	def div(self):
		if self.snd == 0:
			return "Error: Division by zero"
		result = self.fst / self.snd
		return int(result)
```

```python
a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 8)

print(a.add())
print(a.mul())
print(a.sub())
print(a.div())

print(b.add())
print(b.mul())
print(b.sub())
print(b.div())
```

## 생성자
***

```python
a = FourCal()
a.add()
```

```python
class FourCal:
	def __init__ (self, fst, snd):
		self.fst = fst
		self.snd = snd

	def setdata(self, fst, snd):
		self.fst = fst
		self.snd = snd

	def add(self):
		result = self.fst + self.snd
		return result
	
	def sub(self):
		result = self.fst - self.snd
		return result
	
	def mul(self):
		result = self.fst * self.snd
		return result
	
	def div(self):
		if self.snd == 0:
			return "Error: Division by zero"
		result = self.fst / self.snd
		return result
```

```python
a = FourCal()
a.add()
```

```python
a = FourCal(4, 2)
print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
```

## 클래스 상속
***

```python
class MoreFourcal(FourCal):
	def pow(self):
		result = self.fst ** self.snd
		return result
```

```python
a = MoreFourcal(4, 2)
print(a.add())
print(a.pow())	
```

## 메소드 오버라이딩
***

```python
a = FourCal(4, 0)
print(a.div())
```

```python
class SafeFourCal(FourCal):
	def div(self):
		if self.snd == 0:
			return 0
		else:
			return self.fst / self.snd
```

```python
a = SafeFourCal(4, 0)
print(a.div())
```

### 오류 예외처리 기법

```python
# try-except.py
try:
	4/ 0
except ZeroDivisionError as e:
	print(e)
```

#### <span style="color:darkorange">여러 개의 오류 처리하기</span>

```python
# manyerror.py
try:
	a = [1, 2]
	print(a[3])
	4 / 0
except ZeroDivisionError as e:
	print("0으로 나눌 수 없습니다.")
	print(e)
except IndexError as e:
	print("인덱스를 벗어났습니다.")
	print(e)
```

```python
try:
	a = [1, 2, 3]
	print(a[3])
	4 / 0
except (IndexError, ZeroDivisionError) as e:
	print(e)
```

#### <span style="color:darkorange">try-else 문 <spane>

```python
# try-else.py
try:
	age = int(input("나이를 입력하세요: "))
except:
	print("입력이 올바르지 않습니다.")
else:
	if age < 18:
		print("미성년자입니다.")
	else:
		print("성인입니다.")
```

### 예외 만들기

```python
# errormake.py
class MyError(Exception):
	pass

def say_nick(nick):
	if nick == "바보":
		raise MyError()
	print(nick)
```

```python
say_nick("천재")
say_nick("바보")
```

```python
try:
	say_nick("천재")
	say_nick("바보")
except MyError:
	print("허용되지 않는 별명입니다.")
```

```python
try:
	say_nick("천재")
	say_nick("바보")
except MyError as e:
	print(e)
```

```python
class MyError(Exception):
	def __str__ (self):
		return "허용되지 않는 별명입니다."
```

```python
try:
	say_nick("천재")
	say_nick("바보")
except MyError as e:
	print(e)
```
